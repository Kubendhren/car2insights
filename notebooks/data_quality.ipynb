```python
import pandas as pd
import os


data_path ="../data/raw/"

files = os.listdir(data_path)

print("Files available:")
for file in files:
    print(file)
    
customers = pd.read_csv(
    data_path + "olist_customers_dataset.csv"
)

orders = pd.read_csv(
    data_path + "olist_orders_dataset.csv"
)

order_items = pd.read_csv(
    data_path + "olist_order_items_dataset.csv"
)

order_payments = pd.read_csv(
    data_path + "olist_order_payments_dataset.csv"
)

order_reviews = pd.read_csv(
    data_path + "olist_order_reviews_dataset.csv"
)

products = pd.read_csv(
    data_path + "olist_products_dataset.csv"
)

sellers = pd.read_csv(
    data_path + "olist_sellers_dataset.csv"
)

product_category_translation = pd.read_csv(
    data_path + "product_category_name_translation.csv"
)

geolocation = pd.read_csv(
    data_path + "olist_geolocation_dataset.csv"
)


tables = {
    "CUSTOMERS": customers,
    "ORDERS": orders,
    "ORDER_ITEMS": order_items,
    "ORDER_PAYMENTS": order_payments,
    "ORDER_REVIEWS": order_reviews,
    "PRODUCTS": products,
    "SELLERS": sellers,
    "PRODUCT_CATEGORY_TRANSLATION": product_category_translation,
    "GEOLOCATION": geolocation
}

for table_name, df in tables.items():
    print(f"{table_name}: {df.shape}")

for table_name, df in tables.items():
    
    print("=" * 70)
    print(table_name)
    print("=" * 70)
    
    print(df.columns.tolist())
    print()

for table_name, df in tables.items():
    
    print("=" * 70)
    print(table_name)
    print("=" * 70)
    
    print(df.dtypes)
    print()

summary = []

for table_name, df in tables.items():
    
    for column in df.columns:
        
        summary.append({
            "Table": table_name,
            "Column": column,
            "Data_Type": str(df[column].dtype),
            "Rows": df.shape[0],
            "Columns_Count": df.shape[1]
        })

data_dictionary = pd.DataFrame(summary)

data_dictionary

key_information = {
    
    "CUSTOMERS": {
        "customer_id": "PK",
        "customer_zip_code_prefix": "FK"
    },
    
    "ORDERS": {
        "order_id": "PK",
        "customer_id": "FK"
    },
    
    "ORDER_ITEMS": {
        "order_id": "FK",
        "order_item_id": "PK",
        "product_id": "FK",
        "seller_id": "FK"
    },
    
    "ORDER_PAYMENTS": {
        "order_id": "FK",
        "payment_sequential": "PK"
    },
    
    "ORDER_REVIEWS": {
        "review_id": "PK",
        "order_id": "FK"
    },
    
    "PRODUCTS": {
        "product_id": "PK",
        "product_category_name": "FK"
    },
    
    "SELLERS": {
        "seller_id": "PK",
        "seller_zip_code_prefix": "FK"
    },
    
    "PRODUCT_CATEGORY_TRANSLATION": {
        "product_category_name": "PK"
    },
    
    "GEOLOCATION": {
        "geolocation_zip_code_prefix": "PK"
    }
}

dictionary = []

for table_name, df in tables.items():
    
    for column in df.columns:
        
        key = key_information.get(
            table_name, {}
        ).get(column, "")
        
        dictionary.append({
            "Table": table_name,
            "Column": column,
            "Data Type": str(df[column].dtype),
            "Key": key
        })

data_dictionary = pd.DataFrame(dictionary)

data_dictionary

for table_name, df in tables.items():
    
    print("=" * 70)
    print(table_name)
    print("=" * 70)
    
    display(df.head())


for table_name, df in tables.items():
    
    print("=" * 80)
    print(f"TABLE: {table_name}")
    print("=" * 80)
    
    print("Shape:", df.shape)
    
    print("\nColumns:")
    print(df.columns.tolist())
    
    print("\nData Types:")
    print(df.dtypes)
    
    print("\nFirst 3 Records:")
    display(df.head(3))
    
    print("\n")

data_dictionary.to_csv(
    "../output/data_dictionary.csv",
    index=False
)

os.makedirs("../output", exist_ok=True)

data_dictionary.to_csv(
    "../output/data_dictionary.csv",
    index=False
)

```

```python

missing_summary = []

for table_name, df in tables.items():

    for column in df.columns:

        missing_count = df[column].isnull().sum()
        total_count = len(df)
        missing_percentage = (missing_count / total_count) * 100

        missing_summary.append({
            "Table": table_name,
            "Column": column,
            "Missing_Count": missing_count,
            "Total_Rows": total_count,
            "Missing_Percentage": round(missing_percentage, 2)
        })

missing_df = pd.DataFrame(missing_summary)

missing_df
```

```python
duplicate_summary = []

for table_name, df in tables.items():

    duplicate_count = df.duplicated().sum()
    total_rows = len(df)

    duplicate_summary.append({
        "Table": table_name,
        "Total_Rows": total_rows,
        "Duplicate_Rows": duplicate_count,
        "Duplicate_Percentage": round(
            (duplicate_count / total_rows) * 100, 2
        )
    })

duplicate_df = pd.DataFrame(duplicate_summary)

duplicate_df
```

```python
datatype_summary = []

for table_name, df in tables.items():

    for column in df.columns:

        datatype_summary.append({
            "Table": table_name,
            "Column": column,
            "Data_Type": str(df[column].dtype)
        })

datatype_df = pd.DataFrame(datatype_summary)

datatype_df
```

```python
primary_keys = {
    "CUSTOMERS": ["customer_id"],
    "ORDERS": ["order_id"],
    "ORDER_ITEMS": ["order_item_id"],
    "ORDER_PAYMENTS": ["payment_sequential"],
    "ORDER_REVIEWS": ["review_id"],
    "PRODUCTS": ["product_id"],
    "SELLERS": ["seller_id"],
    "PRODUCT_CATEGORY_TRANSLATION": ["product_category_name"],
    "GEOLOCATION": ["geolocation_zip_code_prefix"]
}
```

```python
pk_summary = []

for table_name, columns in primary_keys.items():

    df = tables[table_name]

    for column in columns:

        duplicate_count = df[column].duplicated().sum()
        null_count = df[column].isnull().sum()
        total_rows = len(df)

        pk_summary.append({
            "Table": table_name,
            "Primary_Key": column,
            "Total_Rows": total_rows,
            "Null_Count": null_count,
            "Duplicate_Count": duplicate_count,
            "Is_Unique": duplicate_count == 0,
            "Has_Null": null_count > 0
        })

pk_df = pd.DataFrame(pk_summary)

pk_df
```

```python
def find_outliers(df, column):

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    return outliers
```

```python
price_outliers = find_outliers(
    order_items,
    "price"
)

print("Price outliers:", len(price_outliers))
```

```python
freight_outliers = find_outliers(
    order_items,
    "freight_value"
)

print("Freight value outliers:", len(freight_outliers))
```
